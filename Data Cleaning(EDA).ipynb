# Text Normalization & Categorical Cleaning

```python
df['Gender'].unique()

df['Gender'] = df['Gender'].str.lower().str.strip()
df['Gender'].replace({'m': 'male', 'f': 'female'}, inplace=True)
df['Gender'].unique()
df['City'] = df['City'].str.strip().str.lower()
df['Country'] = df['Country'].str.strip().str.lower()
```

```python
df['Country'].replace({'ind': 'india'}, inplace=True)
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.drop_duplicates(subset=['Customer_ID'], keep = 'first',inplace=True)
```

```python
df['Signup_Date'] = pd.to_datetime(df['Signup_Date'])
df['Last_Purchase_Date'] = pd.to_datetime(df['Last_Purchase_Date'])
```

# Handling Outliers (Age & Purchase Amount)

```python
df.loc[(df['Age'] < 15) | (df['Age'] > 100), 'Age'] = None
df['Age'] = df['Age'].fillna(df['Age'].median()).astype(int)
```

```python
df = df[df['Purchase_Amount'] > 0].copy()

Q1 = df['Purchase_Amount'].quantile(0.25)
Q3 = df['Purchase_Amount'].quantile(0.75)
IQR = Q3 - Q1
upper_limit = Q3 + 1.5 * IQR

df['Purchase_Amount'] = df['Purchase_Amount'].clip(upper=upper_limit)
```

# Logical Consistency Checks (Dates & Constraints)

```python
invalid_dates = df[df['Last_Purchase_Date'] < df['Signup_Date']]
print(f"Invalid Date Records: {len(invalid_dates)}")

df = df[df['Last_Purchase_Date'] >= df['Signup_Date']]
```

# Exporting Cleaned Dataset

```python
df.to_csv("cleaned_customer_sales_data.csv", index=False)
```
